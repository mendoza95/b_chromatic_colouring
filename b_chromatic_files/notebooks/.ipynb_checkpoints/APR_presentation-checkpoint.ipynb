## Colouring problem
Given a graph $G=(V,E)$, assing colours to the vertices of $G$ such that no two adjacent vertices get the same colour

```python
G = nx.Graph()
G.add_nodes_from(["v{}".format(i+1) for i in range(8)])
G.add_edge("v1", "v2")
G.add_edge("v1", "v5")
G.add_edge("v1", "v4")
G.add_edge("v1", "v8")
G.add_edge("v2", "v3")
G.add_edge("v2", "v5")
G.add_edge("v2", "v6")
G.add_edge("v3", "v4")
G.add_edge("v3", "v6")
G.add_edge("v3", "v7")
G.add_edge("v4", "v7")
G.add_edge("v4", "v8")
pos = {"v1":(4,8), "v2":(8,8), "v3":(8,4), "v4":(4,4),
       "v5":(6,12), "v6":(12,6), "v7":(6,0), "v8":(0,6)}
```

```python
nx.draw(G,pos, with_labels=True)
```

The minimum integer $k$ such that $G$ admits a proper colouring is the *chromatic number* of $G$ and it is denoted by $\chi(G)$.

## Partition redistribution heuristic
An interesting way of approximating a colouring using $\chi(G)$ colours is the partition redistribution heuristic:
- Start with a colouring $\mathbf{c}:V\rightarrow C$ of $G$ using $k'>\chi(G)$ colours
- For every colour $c\in C$:
    - For every vertex $v\in V(G)$ of colour $c$:
        - If there exists a colour $c'$ that is not in the neighbourhood of $v$ then assign $\mathbf{c}(v)=c'$
    - If there is no vertex of colour $c$ then we can remover $c$ from $C$
    
This algorithm removes one colour at a time trying to minimize the number of colours used. 

If there is no colour $c$ to be removed then this implies that for every colour $c\in C$ there exists a vertex $v$ such that is adjacent to vertices of every other colour.

```python
f2 = {u:i for i, u in enumerate(G.nodes())}
f2 = partition_redistribution1(G, f2)
```

## Partition redistribution heuristic

Example of the partition redistribution on the graph $G$

```python
colours = get_colors_to_use(f2)
nx.draw(G, pos, with_labels=True, node_color=colours)
```

Notice that 4 colours were used while $\chi(G)=3$

## b-chromatic number of a graph
The worst-case behaviour of the *partitiion redistribution* heuristic give rise to another graph colouring parameter: the *b-chromatic number* of a graph.

The *b-chromatic number* of a graph $G$ is the maximum integer $k$ for which $G$ admits a colouring $\mathbf{c}:V(G)\rightarrow C$ such that for every colour $c\in C$ there exists a vertex $v\in V(G)$ that is adjacent to vertex of every other colour.

The *b-chromatic number* of $G$ is denoted by $\varphi(G)$. 

```python
nx.draw(G, pos, with_labels=True, node_color=colours)
```

$\varphi(G)=4$.
