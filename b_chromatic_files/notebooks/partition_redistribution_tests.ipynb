#### We set a max number of iteration to find the best value

```python
N = 1000
```

#### Function for testing a PR algorithm N times and return the best result and computation time

```python
def run_test(G, f, pr_function, N):
    """ Runs a partition redistribution algorithm N times and return the best result and computation time
        G: Graph,
        f: Colour assignment function,
        pr_function: Partition redistribution function
        N: Number of iterations
    """
    best_phiG = 0
    start = time.time()
    for _ in range(N):
        fp = pr_function(G, f)
        Cp = set(fp.values())
        Vp = get_b_chromatic_vertices(G, fp)
        assert is_proper(G, fp)
        assert are_b_chromatic(G, fp, Vp)
        if len(Cp) > best_phiG: best_phiG = len(Cp)
    end = time.time()
    return best_phiG, end-start
```

### Partition Redistribution 1

```python
best_phiG, total_time = run_test(G, f, partition_redistribution1, N)
print("Best approximation over {} iterations: {}".format(N, best_phiG))
print("Total time: {} s".format(total_time))
```

### Partition Redistribution 2

```python
best_phiG, total_time = run_test(G, f, partition_redistribution2, N)
print("Best approximation over {} iterations: {}".format(N, best_phiG))
print("Total time: {} s".format(total_time))
```

### Partition Redistribution 3

## Testing over multiple random graphs with different densities

```python
N = 500
results = {"#vertices":[], "density":[], "DELTA(G)": [], "m(G)":[],
           "PR1(G)":[], "TPR1(s)":[], "PR2(G)":[], "TPR2(s)":[], "PR3(G)":[], "TPR3(s)":[]}
graph_sizes_list = [n for n in range(20, 110, 10)] + [n for n in range(150, 350, 50)]
density_values = [0.2, 0.4, 0.6, 0.8]
```

```python
for gsize in graph_sizes_list:
    for d in density_values:
        G = nx.random_graphs.gnp_random_graph(gsize, d)
        f = {u:i+1 for i, u in enumerate(G.nodes())}
        mG = get_m_degree(G)
        DELTA = max([G.degree(u) for u in G])
        results["#vertices"].append(G.number_of_nodes())
        results["density"].append(d)
        results["DELTA(G)"].append(DELTA)
        results["m(G)"].append(mG)
        
        best_phiG, total_time = run_test(G, f, partition_redistribution1, N)
        results["PR1(G)"].append(best_phiG)
        results["TPR1(s)"].append(total_time)

        best_phiG, total_time = run_test(G, f, partition_redistribution2, N)
        results["PR2(G)"].append(best_phiG)
        results["TPR2(s)"].append(total_time)

        best_phiG, total_time = run_test(G, f, partition_redistribution3, N)
        results["PR3(G)"].append(best_phiG)
        results["TPR3(s)"].append(total_time)
```
