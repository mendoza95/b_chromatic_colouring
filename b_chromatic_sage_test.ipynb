```python
from sage.graphs.all import graphs
from sage.graphs.graph_coloring import b_coloring
import warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)
```

```python
def get_m_degree(G):
    n = len(G.vertices())
    X = [0 for _ in range(0, n+1)]
    Z = [0 for _ in range(0, n+1)]
    max_j = 0
    for u in G.vertices(): X[G.degree(u)] += 1
    for j in range(n-1, 0, -1): Z[j] = Z[j+1] + X[j]
    for j in range(1, n):
        if Z[j-1] >= j:
            max_j = j
    return max_j
```

__Claim__: Let $G=(V,E)$ be a $d$-regular graph with $n=|V|$ vertices. If $d+2\leq n\leq 2d+1$ then $\varphi(G)=d$.

```python
def show_colourings(d):
    results = {}
    for n in range(d+2, 2*d+2, 1):
        G = graphs.RandomRegular(d, n)
        if G is not False:
            results[n] = {}
            m = get_m_degree(G)
            b, coloring = b_coloring(G, m, value_only=False)
            results[n]["m-degree"] = m
            results[n]["b-chromatic"] = b
            results[n]["coloring"] = coloring
            print("{}-regular graph on {} vertices with m(G)={} and b-chromatic number: {}".format(d, n, m, b))
            print(coloring)
            results[n]["adjacent_list"] = {}
            for u in G.vertices():
                print("{} -> {}".format(u, G.neighbors(u)))
                results[n]["adjacent_list"][u] = G.neighbors(u)
            print("\n")
    return results
```
